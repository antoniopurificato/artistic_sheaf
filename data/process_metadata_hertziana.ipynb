```python
import re
import numpy as np
import pandas as pd


# --- 2) Normalize placeholders like '-' -> NaN (especially medium_short, but safe across the DF) ---
df = df.applymap(lambda x: x.strip() if isinstance(x, str) else x)
df.replace({"-": np.nan, "": np.nan}, inplace=True)

# --- 3) Parse a year from the "date" column ---
YEAR_RE = re.compile(r"(?<!\d)(\d{3,4})(?!\d)")  # matches 3–4 digit standalone numbers

def extract_year(val):
    """
    Tries to extract a representative year from messy date strings like:
      '1953/1956', 'vor 1958', '2009.10.20', '1044'
    Strategy:
      - find all 3–4 digit numbers
      - return the *earliest* year found (min), as a conservative representative
    """
    if pd.isna(val):
        return np.nan
    s = str(val)
    years = [int(m.group(1)) for m in YEAR_RE.finditer(s)]
    if not years:
        return np.nan
    return min(years)

df["year"] = df["date"].map(extract_year).astype("Float64")  # nullable integer-like

# --- 4) Assign 50-year periods ---
def year_to_50yr_period(y):
    """
    Example: 1044 -> '1000-1049'
    """
    if pd.isna(y):
        return pd.NA
    y = int(y)
    start = (y // 50) * 50
    end = start + 49
    return f"{start}-{end}"

df["period_50yr"] = df["year"].map(year_to_50yr_period).astype("string")

# --- 5) Translate medium_short to English via mapping ---
medium_map = {
    # Painting-related
    "Malerei": "Painting",
    "Tafelmalerei": "Panel painting",
    "Deckenmalerei": "Ceiling painting",
    "Wandmalerei": "Mural painting",
    "Gemälde": "Painting",

    # Sculpture
    "Skulptur": "Sculpture",
    "Altarskulptur": "Altarpiece sculpture",
    "Grabskulptur": "Tomb sculpture",

    # Drawing / Graphic arts
    "Zeichenkunst": "Drawing",
    "vorbereitende Zeichnung": "Preparatory drawing",
    "Druckgraphik": "Printmaking",

    # Book / Script arts
    "Buchkunst": "Book art",
    "Schriftkunst": "Calligraphy",
    "Handschrift": "Manuscript",

    # Applied / Decorative
    "Angewandte Kunst": "Applied arts",
    "Kunstgewerbe": "Decorative arts",

    # Architecture
    "Architektur": "Architecture",

}
# ensure '-' already removed; still guard just in case
df["medium_short_en"] = (
    df["medium_short"]
    .replace({"-": np.nan})
    .map(medium_map)
    .fillna(df["medium_short"])  # keep original if something unexpected appears
)
```
